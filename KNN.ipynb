```python
import numpy as np
from collections import Counter

class KNN:
    def __init__(self,k=3):
        self.k=k

    def fit(self,X,y):
        self.X_train=np.array(X)
        self.y_train=np.array(y)

    def euclidean_dist(self,point_a,point_b):
        return np.linalg.norm(point_a-point_b)

    def classify_point(self,distances):
        sorted_distances=sorted(distances,key=lambda x:x[0])
        top_k_sorted_distances = sorted_distances[:self.k]
        labels=[item[1] for item in top_k_sorted_distances]
        most_common_label = Counter(labels).most_common(1)[0][0]

        return most_common_label
    

    def predict(self,X):
        self.X_test=np.array(X)
        y_predict=[]
        for sample in self.X_test:
            distances=[]
            for train_sample, label in zip(self.X_train, self.y_train):
                distances.append((self.euclidean_dist(sample, train_sample), label))

            y_predict.append(self.classify_point(distances))

        return y_predict
            



```

```python
X_train = [[1, 2], [2, 3], [3, 1], [6, 5], [7, 7], [8, 6]]
y_train = ['A', 'A', 'A', 'B', 'B', 'B']

# New points to classify
X_test = [[2, 2], [7, 6]]

knn = KNN(k=3)
knn.fit(X_train, y_train)

print("Predictions:", knn.predict(X_test))
```

```python
from sklearn.datasets import load_iris
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.neighbors import KNeighborsClassifier
from sklearn.metrics import accuracy_score

# 1. Load dataset & split
data = load_iris()
X_train, X_test, y_train, y_test = train_test_split(
    data.data, data.target, test_size=0.2, random_state=42
)

# 2. Scale features (essential for distance-based models like KNN)
scaler = StandardScaler()
X_train = scaler.fit_transform(X_train)
X_test = scaler.transform(X_test)

# 3. Test your custom KNN
my_knn = KNN(k=3)
my_knn.fit(X_train, y_train)
my_preds = my_knn.predict(X_test)

# 4. Compare against scikit-learn
sk_knn = KNeighborsClassifier(n_neighbors=3)
sk_knn.fit(X_train, y_train)
sk_preds = sk_knn.predict(X_test)

print(f"Your KNN Accuracy:    {accuracy_score(y_test, my_preds):.4f}")
print(f"Sklearn Accuracy:     {accuracy_score(y_test, sk_preds):.4f}")
print(f"Predictions match:    {np.array_equal(my_preds, sk_preds)}")
```
